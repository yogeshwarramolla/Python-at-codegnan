```python
# round(val, position)
print(round(10.25546,2))
print(round(10.25633,2))
print(round(10.25342,2))
```

```python
#chr() : returns the "character" when ASCII value entered
#ord() : returns the "ASCII" when charc enetered
print(chr(99))
print(chr(65))
print(ord('B'))
```

```python
#all()
#any()
print(all([1,2,3]))
print(any([1,0,0,0]))
print(all([False,False]))
```

```python
#isinstance(obj,class_name): checks whether the object belongs to the particular class_name
print(isinstance(11,int))
print(isinstance(12.9,int))
print(isinstance((1,2,3),tuple))
```
