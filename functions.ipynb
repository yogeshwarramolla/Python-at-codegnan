```python
def add(x:int , y:int):
    c=x + y
    return c

#function calling
a,b=10,20
res=add(a,b)
print(res)
print(add(374,47))
      
```

```python
#with parameters without return : so return is none 
def add(x:int , y:int):
    c=x + y
    return

#function calling
a,b=10,20
res=add(a,b)
print(res)
```

```python
#without paramters with return 
def add():
    x=10
    y=20
    c=x + y
    return c

#function calling

res=add()
print(res)
```

```python
#without para ,without arguments: returns the none value
def add():
    c=x + y
    return c

#function calling

res=add()
print(res)
```

```python
def even(x:int):
   if x % 2==0:
       return"even"
   else:
       return"Odd"
           
a=20
res=even(a)
print(res)
```

```python
def prime(x:int):
   if x % 1==0 and x % x==0:
       return"prime"
   else:
       return"not"
           
a=int(input())
res=prime(a)
print(res)
```

```python
def prime(x):
    for i in range(2, x):
        if x % i == 0:
            return "Not Prime"

    return "Prime"

a = int(input())
res = prime(a)
print(res)
```
