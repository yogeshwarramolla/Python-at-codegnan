```python
def prime(x):
    for i in range(2, x):
        if x % i == 0:
            return "Not Prime"

    return "Prime"

a = int(input())
res = prime(a)
print(res)
```

# types of arguments
   - postional arguments

```python
def sub(x:int , y:int):
    z=x-y
    return z
a,b=20,10
print(sub(a,b))
print(sub(b,a)) # change pos
```

 - keyword arguments : arguments are explicitively passed with parameter names

```python
def sub(x:int , y:int):
    z=x+y
    return z
a,b=20,10
print(sub(x=a,y=b))
print(sub(y=b,x=a)) 
```

```python
def sub(x:int ,y:int,w:int):
    z=x+y+w
    return z
a,b,c=map(int,input().split())
print(sub(x=a,y=b,w=c))
#print(sub(y=b,x=a ,w=c))
#print(4,y=10,w=30)  # this doesn't work. why?
```

 - default arguments: we assign the defalut values to the parameters , they can be reassigned in the calling function

```python
def sub(x:int ,y:int=0,w:int=0):
    z=x+y+w
    return z
a,b,c=map(int,input().split())
print(sub(a,b))
print(sub(a,b,c)) # usually if we doesn't involve all the variables we will get the error but here we won't get the error
```

```python
def sub(x:int ,y:int=0,w:int=1):
    z=x+y*w
    return z
a,b,c=map(int,input().split())
print(sub(a,b))
print(sub(a,b,c)) # usually if we doesn't involve all the variables we will get the error but here we won't get the error
```

- variable length positional arguments : we can assign the number values at a time using *args , note: the order is pos , default and variable len argts

```python
def sub(x ,y=0, *args):
    print(type(args))
    s=0
    for num in args:
        s+=num
    return s
print(sub(10 , 2 , 3 ,4 , 5))
```

- variable length keyword arguments : dictionary as input

```python
def sub(x ,y=0, *args , **kwargs):
    print(type(kwargs))
    print(kwargs)
    print(kwargs.values())
    print(kwargs.keys())
print(sub(10 , 2 , 3 ,4 , 5, name ="yogi",Age=20))
```
