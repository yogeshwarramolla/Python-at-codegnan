# string to int, float, bool,list,tuple,set 
# if the string is a number value

```python
s='123456'
print(int(s))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```

# if string has float numbers

```python
s='123456.3424'
print(int(float(s)))
print(float(s))
print(bool(s))
print(list(s))
print(tuple(s))
print(set(s))
```
