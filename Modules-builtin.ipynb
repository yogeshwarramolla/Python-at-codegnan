# RANDOM

```python
import random
print(random.random()) # returns in between of 0 and 1
```

```python
print(random.randint(2,10)) # returns a digit between the alloted
print(random.randint(14329,13243424))
```

# A game 
  - who will reach the 100 

```python
count=rolls=0
while True:
    input("To throw a dice, click enter")
    face_val=random.randint(1,6)

    if count+face_val<=100:
        count+=face_val
    rolls+=1
    if count==100:
        print(f"you reached the top in {rolls} chances")
        break
    print(f"your Dice value:{face_val}")
    print(f"your at {count} position in {rolls} chances")
    print("*"*50)
```
