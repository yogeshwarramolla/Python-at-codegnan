```python
#set operations
#union(set):all elememnts , unique 
# intersection(set): all the common elements
a1={1,2,3,4}
a2={2,3,6,7}
a3={1,2,3,4,5,4,4}
uni=a1.union(a2 , a3) # union
print(uni)
res= a1 | a2 | a3 # mathematically
print(res)
inter= a1.intersection(a2 , a3) # intersection
print(inter)
inter1= a1 & a2 & a3 # mathematically
print(inter1)
dif=a1.difference(a2 , a3) # difference 
print(dif)
dif1= a1 - a2 - a3
print(dif1)
print(a1 ^ a2) # symmetric difference

```

```python
# issubset()
#issuperset()
#isdisjoint () all the aboce gives "true or false"
a={1,2,3}
b={1,2,3,4,5}
print(a.issubset(b))
print(a.issuperset(b))
print(a.isdisjoint(b))
```

```python
# usually intersection, difference doesn't change the original adress but the intersection_update $ difference_update change the address 
#so we get change in set
a1={1,2,3,4}
a2={2,3,6,7}
a3={1,2,3,4,5,4,4}
a1.intersection_update(a2 , a3) # only the first set will change remaining same
print(a1,a2,a3)
```
