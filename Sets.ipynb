# Set Characteristics

```python
# set uordered collection
# set doesn't allow duplicate , only unique elements
a={1, 2,3,3,2,3,"hello",2.04}
print(a)
```

```python
#no indexing in set
a={1, 2,3,3,2,3,"hello",2.04}
b=a[0]
print(b)
```

```python

#list can't be part of sets
a={1, 2,3,3,2,3,"hello",[2.04,3.5]}
print(a)
```
