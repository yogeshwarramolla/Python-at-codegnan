```python
#generate the bill morethan 500 give 10% discount otherwise give 0% discount
bill= int(input())
if bill > 500:
    discount=bill -(bill * 0.1) 
    print(f"after disount:{discount}")
else:
    print(f"the bill without discount: {bill}")
```

```python
products = {'pen':5 , 'book': 20 ,'pencil':3}
item=input().lower()
quantity= int(input())
if item in products:
    total_amount=products.get(item)* quantity
    print(f"{quantity} of {item} are : {total_amount}")
else: 
    print("not available")
```
