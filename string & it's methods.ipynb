# .index(): to get the substring index place of the string
# .find(): returns -1 if it's beyond the indexing

```python
b= " hello namste ths is yogi"
print(b.index("ths"))
print(b.find('z'))
print(b.find("o",7)) # here 7 means chekcs the letter after 7th position
```

# .count() : used to get the count of the non - overlapping occurences 

```python
a="ABCBCBCbbaab"
print(a.count('BCB'))
print(a.count('B') + a.count('b'))
```

# .replace (old str , new str)

```python
b= " hello namste ths is yogi"
print(b.replace("ths","THIS"))
print(b.replace('o', '2',2))
# 2 means the first two occurences                
```

## string checking methods
- returns boolean
 - isupper(): returns T when the whole str in caps
 - islower(): F if all not in caps
 - isalpha(): T , if all contains alphabets
 - isdigit(): T, all digits
 - isalnum(): T , both alpha & digits
 - startswith(substring): T , if stsrts with substr
 - endswith(substring): T

```python
print('YOGI'.isupper())
print('yogi'.islower())
print('123','yogi'.isalpha())
print('123'.isdigit())
print('234yogi'.isalnum())
a= 'hey man'
print(a.startswith('hey') + a.endswith('man'))
```
