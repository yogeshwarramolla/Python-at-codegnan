```python
def tozero():
    n=int(input())
    c=0
    while n>0:
        if n % 2 == 0:
            n//=2
            c+=1
        else:
            n-=1
            c+=1
    return c
    
tozero()    
```

```python
class Solution(object):
    def diagonalSum(self, mat):
        sum=0
        for i in range(len(mat)):
            for j in range(len(mat)):
                  if i==j or j==len(mat)-i-1:
                       sum+=mat[i][j]
        return sum     
                
```

```python
class Solution(object):
    def numJewelsInStones(self, jewels, stones):
        a=len(jewels)
        b=len(stones)
        c=0
        for i in range(a):
            for j in range (b):
                if jewels[i]==stones[j]:
                    c+=1
        return c  
#571
```
