# Dictionary Methods
   - .get(key, default_values): returns the key's value if exsists , if not error won't appear

```python
Details = { 'Name':'yogi','course':'python','age':22}
b=Details.get('Name')
c=Details.get("Batch", "not avilable")
print(b)
print(c)
```

# .update(new dictionary) : used to update the dic with the new keys also to change the pre exisiting key's values

```python
Details = {'Name':'yogi','course':'python','age':22}
Details.update({'age':20 , 'village' : 'kopperga'})
print(Details)
```

# - keys(): returns all the list of keys in the dic
# - values(): only retutns the list of values  part
# - items(): list of tuple of key-value pairs

```python
Details = {'Name':'yogi','course':'python','age':22}
a=Details.keys()
b=Details.values()
c=Details.items()
print(a)
print(b)
print(c)
```

# remove methods
 - popitem(): removes the last key-value pair and returns the tuple as tuple value
 - pop(key): removes the specific the key- value pair , orelse key error

```python
Details = {'Name':'yogi','course':'python','age':22}
a=Details.popitem()
print(a)
Details.pop('course')
print(Details)
```
