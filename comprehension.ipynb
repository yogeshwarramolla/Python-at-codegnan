```python
n=int(input())
res=[x*2 for x in range(0,n)]
print(res)
```

```python
n,m=map(int,input().split())
res=[x  for  x in range(n,m+1) if x%2==1]
print(res)
```

```python
arr = list(map(int, input().split()))

res = [x**2 if x % 2 == 0 else round(x**0.5 , 2) for x in arr]

print(res)
```

```python
res = [(x,y) for x in range(1,7) for y in range(1,7)]
print(res)
```

```python
res=[(x,y) for x in range(1,7) for y in range(1,7) if x % 3 == 0]
print(res)
```

```python
st=input()

res=[char for char in st.lower() if char in "aeiou"]
print(res)
print(len(res))
```

```python

#set comprehension 
arr = list(map(int, input().split()))

res = {x**2 if x % 2 == 0 else round(x**0.5 , 2) for x in arr}
print(type(res))
print(res)
```

```python
arr = list(map(int, input().split()))

res = {x:x**2 if x % 2 == 0 else round(x**0.5 , 2) for x in arr}
print(type(res))
print(res)
```
