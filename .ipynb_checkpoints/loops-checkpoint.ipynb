# for loop syntax:
   - for iterator in iterable:
        #block of code

```python
a = int(input())

for i in range(1, a + 1):
    if i % 2 == 0:
        print(f"{i} is even")
    else:
        print(f"{i} is odd")
        
        
      
```

```python
a=input().lower()
c=0
v=0
for char in a:
    if char.isalpha():
        if char in "a,e,i,o,u":
          v+=1
    else: 
        c+=1
print(f"vowels in {a} are {v}")
print(f"con in {a} are {c}")
```

```python
income = 5000
total = 0

for month in range(1, 7):
    print(f"Month {month} : {income}")
    total += income
    income += 1000

print(f"\nTotal Income = {total}")
```
