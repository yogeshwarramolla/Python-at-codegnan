# List : ordered , mutable, [],indexing & slicing , heterogenous elements

```python
s=[1,2,6,"yogi",6,19]
print(s)
print(s[1]+ s[2])
print(s[3:4:2])
print(s[:4:-1])
```

```python
a=[1,12,23,16]
b=[34,56,78]
print(a + b)
print(a[1]*2)
print(a*2)
```

# Item adding methods
  - append(item): add item at the end
  - extend(item): adds multiple items at the end of the list
  - insert(pos, item)

```python
lst=[10,20,34]
lst.append(10000)
print(lst)
lst.extend((400,500))
print(lst)
lst.insert(4,(50, 60))
print(lst)
```

# item removing methods
  - remove (item)
  - pop(pos): index is used
  - clear(list): everything will be removed

```python
a=[10,46,89]
a.remove(46)
print(a)
a.pop(1)
print(a)
a.pop()
print(a)
```

# index and count
  - index(item): returns index value
  - count(item):returns the count of the particular item

```python
lst=[10,2,3,2,2,2,40]
a=lst.index(2) # here i assigned a new variable so original doesn't change
print(a)
b=lst.count(2)
print(b)
```

# sorting
  - sort(reverse = False) : default false ascending order , reverse = True descending order

```python
st=[10,2,3,2,2,2,40]
st.sort()
print(st)
st.sort(reverse = True) # we can't assign as a=st.sort() it gives none
print(st)
```
