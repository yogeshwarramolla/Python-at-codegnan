# Scope of Variable

```python
x=10
def out():
    y=20
    def inner():
        z=30
        print(x,y,z)
    inner()
out()    
```

```python
# using global in local scope
x=10
def out():
    y=x+1
    print("first the local is printed",y)
out()
print(x)
```

```python
# we can access the global varible but we can't update it
# using global in local scope
x=10
def out():
    x=x+1
    print("first the local is printed",y)
out()
print(x)
```

```python
# we can update the global in local var by "global" keyword
# note: the global will be updated all along the code , not only in local
x=10
def out():
    global x
    x=x+1
    print(x)
out()
print(x)
```

```python
# we can't update the enclosing in local Directly
# but we can using the "nonlocal"keyword
x=10
def out():
    y=20
    def inner():
        nonlocal y
        z=30
        y+=20
        print(x,y,z)
    inner()
out()    
```

```python
# if the global and the local variables are same but with different values then it will be correspding variable respectively
x=10
def out():
    x=12
    x=x+8
    print(x)
out()   
print(x)
```
