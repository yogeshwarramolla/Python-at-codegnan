# loop control statements
 - continue: After executing of continue keyword execution immediately it goes to next iteration and skips the current iteration after continue keyword statements
 - break: After executing of break keyword immediately execution goes out ofthe loop 
   

```python
ls=list(map(int,input().split()))
target=5
n=len(ls)
for i in range(n):
     if ls[i] == 5:
         print(i)
         break
         
else:
    print(-1)
```

```python
n=int(input())
factors= list()
for i in range (1,n+1):
    if n % i == 0:
        factors.append(i)
        
print(factors)   
        

```
