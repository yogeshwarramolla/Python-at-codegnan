```python
def fact(x):
    if x==0 or x==1:
        return 1
    return x * fact(x-1)
n=int(input())    
print(fact(n))    
```

```python
def sum(x):
    if x==0:
        return 0
    return x % 10 + sum(x//10)   

print(sum(456))
```

```python
def sum(x):
    if x==0:
        return 0
    return 1 + sum(x//10)   

print(sum(456))
```
