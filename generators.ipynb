```python
def num(n):
    for i in range(1,n+1):
        yield i
res=num(10)
print(next(res))
print(next(res))
```

```python
import sys
def num(n):
    for i in range(1,n+1):
        yield i
res=num(10)
print(sys.getsizeof(next(res)))
print(next(res))
```

```python
import time
import sys
s=time.time()
def num(n):
    for i in range(1,n+1):
        yield i**3
        
res=num(2)
print(next(res))
print(next(res))
# to prevent from occuring the erorr , we can use the loop
for val in res:
    print(val)
e=time.time()
print(e-s)
# the list implements faster than the generator but the gen is memeory efficient.
```
